Let the filtered acceleration signal be $a=\begin{bmatrix}4.3 & 2.5 & -8.6\end{bmatrix} m/s^2$.
Then the quaternion for inclination correction as per Laidig's method (from their VQF paper) is $[q_w,q_x,q_y,q_z]=$ ___ , ___ , ____ , ___  // round to 2 decimal digits e.g. 0.01,0.02,0.02.0.04

```python
a = np.array([4.3, 2.5, -8.6])

a_norm = a / np.linalg.norm(a)
a_x, a_y, a_z = a_norm

alpha = np.arccos(a_z)
print(f"{alpha=}")

a_w = np.cos(alpha / 2)
a_x_q = a_x / (2 * a_w)
a_y_q = a_y / (2 * a_w)
a_z_q = 0

q_diff = np.array([a_w, a_x_q, a_y_q, a_z_q])
print(f"{q_diff=}")
q_diff_rounded = np.round(q_diff, 2)
print(f"{q_diff_rounded=}")
```
